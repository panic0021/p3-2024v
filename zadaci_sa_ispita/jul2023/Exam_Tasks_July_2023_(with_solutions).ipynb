#### 1.4 Transformisati kolone u numeričke

#### 1.5 Kreirati kolonu Is_Coffee

Kreirati kolonu `Is_Coffee` koja ce imati vrednost True ukoliko su zadovoljeni sledeći uslovi, u svim ostalim slučajevima ima vrednost False:

* kategorija pića (`Beverage_category`) je **različita** od "Smoothies", "Tazo® Tea Drinks" i "Frappuccino® Blended Crème"
* ukoliko kategorija pića (`Beverage_category`) ima vrednost "Shaken Iced Beverages", vrsta pića (`Beverage`) **ne** sadrži termin "Tazo® Tea"
* ukoliko kategorija pića (`Beverage_category`) ima vrednost "Signature Espresso Drinks", vrsta pića (`Beverage`) **ne** sadrži termine "Hot Chocolate" i "Caramel Apple Spice"

````python
def is_coffee(bev_category, bev_name):
    bev_count = len(bev_category)
    res = [False]*bev_count
    for i, z in enumerate(zip(bev_category, bev_name)):
        cat,name = z
        if cat not in ['Smoothies', 'Tazo® Tea Drinks', "Frappuccino® Blended Crème"]:
            res[i] = True
        if (cat == "Shaken Iced Beverages") and (name.find("Tazo® Tea") >= 0):
            res[i] = False
        if (cat == "Signature Espresso Drinks") and ((name.find("Hot Chocolate") >= 0) or (name.find("Caramel Apple Spice") >= 0)):
            res[i] = False
    return res

data['Is_Coffee'] = is_coffee(data.Beverage_category.to_list(), data.Beverage.to_list())
````

#### 4. Odrediti najviše i najmanje zdrava pića u Starbucks ponudi (13p)

Koja kombinacija vrste pića (`Beverage`) i načina pripreme (`Beverage_prep`) daje pića koja su najmanje zdrava, a koja daje pića koja su najviše zdrava, medju Starbucks pićima? Kao kriterijume za to koliko je piće zdravo uzeti količinu vitamina A i C, šećera, trans masti, holesterola i natrijuma (tj. soli). Konkretno:

* Najzdravijim pićima smatrati ona kod kojih je količina vitamina A i C iznad vrednosti medijane, a količina šećera i trans masti ispod medijane
* Najmanje zdravim smatrati ona pića kod kojih je količina šećera, trans masti, holesterola i soli iznad 75-tog percentila

Za svaku od ove dve grupe, prikazati pića koja im pripadaju i to: pun naziv pića (kombinacija vrste pića i načina pripreme) i vrednosti karakteristika na osnovu kojih su grupe formirane: vitamini A i C, šećer i trans masti za grupu "zdravijih" pića, odnosno šećer, trans masti, holesterol i so za grupu "nezdravih" pića

````python
def concat_bev_data(bev_kind, bev_prep):
    res = []
    for kind, prep in zip(bev_kind,bev_prep):
        res.append(f"{kind} - {prep}")
    return res

data['Beverage_full'] = concat_bev_data(data.Beverage.to_list(), data.Beverage_prep.to_list())
````

### 5. Opšta funkcija za selekciju Starbucks pića na osnovu zadatih kriterijuma (15p)

Napisati funkciju koja za dati proizvoljni niz kriterijuma (zadatat u formi imenovanog ulaznog argumenta proizvoljne dužine (kwargs)), pronalazi i vraća pića koja ispunjavaju date kriterijume. Svaki kriterijum se sastoji od dva elementa: i) naziv varijable (kolone data frame-a) i ii) opseg vrednosti, u formi uređenog para koji čine donja i gornja granica vrednosti varijable. Funkcija vraća data frame koji sadrži sledeće kolone: vrsta pića, oblik pripreme, i atribute iz zadatih kriterijuma.

Primer ulaznog argumenta funckije dat je u nastavku:
```
example_criteria = {'Sugars..g.': (0,10),
                    'Magnesium..mg.': (0,10),
                    'Protein..g.': (5,10),
                    'Iron....DV.': (25, 75)}`
```

Napomena: predvideti da funkciji mogu biti prosleđeni nazivi varijabli koje se ne nalaze među kolonama data frame o Starbucks pićima, kao i da mogu biti prosleđeni opsezi vrednosti koji ne odgovaraju onima u data frame-u. U takvim slučajevima, ispisati poruku greške i ne razmatrati taj kriterijum.


````python
def drink_selection(**criteria):
    col_names = data.columns.to_list()
    data_sub = data.copy()

    vars_to_report = ['Beverage', 'Beverage_prep']

    for var_name, var_range in criteria.items():
        if var_name not in col_names:
            print(f"Unknown variable '{var_name}', will skip it")
            continue

        lower_bound,upper_bound = var_range
        if lower_bound > data_sub[var_name].max() or upper_bound < data_sub[var_name].min():
            print(f"Upper and/or lower bounds are out of range => cannot apply criterion {var_name}")
            continue

        vars_to_report.append(var_name)
        data_sub = data_sub.loc[(data_sub[var_name] > lower_bound) & (data_sub[var_name] < upper_bound),:]

    return data_sub[vars_to_report]
````

````python
example_criteria = {'Sugars..g.': (0,10),
                    'Magnesium..mg.': (0,10),
                    'Protein..g.': (5,10),
                    'Iron....DV.': (25, 75)}

drink_selection(**example_criteria)
````
